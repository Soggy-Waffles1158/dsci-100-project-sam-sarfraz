#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:

```r
tidy_womens_data <- womens_data |> 
  # Keep the columns required for my research question
  select(school, conference, reg_wins, conf_rank, total_wins) |> 
  # Clean conference name inconsistencies and trim whitespace
  mutate(conference = str_trim(conference),conference = case_when(conference == "Trans-America" ~ "Trans America",TRUE ~ conference )) |>
#these two are the same but would populate as different

  # Filter out rows missing total_wins or conference rank 

  filter(!is.na(total_wins), !is.na(conf_rank))

head(tidy_womens_data, 6)
```

```r
data_summary <- tidy_womens_data |> 
  summarise(across(where(is.numeric), 
                   list(mean = ~mean(.x, na.rm = TRUE), 
                        min = ~min(.x, na.rm = TRUE), 
                        max = ~max(.x, na.rm = TRUE))))
data_summary
```

This scatter plot shows the relationship between total wins and regular season wins. There is a clear positive trend. We can see that as teams have more regular season wins they tend to have more total wins. This correlation is somewhat forced as total wins will always contain regular season wins. This will help us answer our research question by showing us how well regular season wins translate to total wins.

```r
plot1 <- ggplot(tidy_womens_data, aes(x = reg_wins, y = total_wins)) +
  geom_point(alpha = 0.6, color = "steelblue") +
  geom_smooth(method = "lm", color = "darkred", se = FALSE) + 
  labs(
    title = "Relationship Between Total Wins and Regular Season Wins",
    x = "Regular Season Wins",y = "Total Wins")

plot1
```
